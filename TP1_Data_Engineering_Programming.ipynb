#  4. Fungsi Menghitung Luas Segitiga

```python
# Fungsi menghitung luas segitiga
# Bisa pakai alas & tinggi, atau 3 sisi (rumus Heron)

def luas_segitiga(alas=None, tinggi=None, a=None, b=None, c=None):
    """
    Menghitung luas segitiga.
    Parameter:
    - alas, tinggi : jika diketahui alas dan tinggi
    - a, b, c : jika diketahui panjang ketiga sisi (rumus Heron)

    Return: luas segitiga (float)
    """
    # Jika pakai alas dan tinggi
    if alas is not None and tinggi is not None:
        luas = 0.5 * alas * tinggi
        cara = "Menggunakan rumus 1/2 * alas * tinggi"

    # Jika pakai rumus Heron
    elif a is not None and b is not None and c is not None:
        s = (a + b + c) / 2  # setengah keliling
        luas = (s * (s - a) * (s - b) * (s - c)) ** 0.5
        cara = "Menggunakan rumus Heron"

    else:
        raise ValueError("Masukkan alas & tinggi ATAU panjang ketiga sisi segitiga.")

    return luas, cara


# Contoh pemakaian di Jupyter/Colab
# Kasus 1: diketahui alas dan tinggi
luas1, metode1 = luas_segitiga(alas=17, tinggi=9)
print(f"Luas segitiga = {luas1:.2f} ( {metode1} )")

# Kasus 2: diketahui ketiga sisi
luas2, metode2 = luas_segitiga(a=2, b=2, c=3)
print(f"Luas segitiga = {luas2:.2f} ( {metode2} )")
```

# 5. Logika conditional dengan untung mengkonversikan suatu nilai dengan aturan

```python
# Fungsi konversi nilai ke huruf
def konversi_nilai(skor):
    """
    Mengonversi skor angka menjadi huruf berdasarkan aturan:
    A ≥ 80
    60 ≤ B < 80
    50 ≤ C < 60
    40 ≤ D < 50
    E < 40
    """
    if skor >= 80:
        return "A"
    elif skor >= 60:
        return "B"
    elif skor >= 50:
        return "C"
    elif skor >= 40:
        return "D"
    else:
        return "E"

# Input nilai dengan bilangan desimal (float)
try:
    nilai = float(input("Masukkan nilai Anda (0-100): "))
    if 0 <= nilai <= 100:
        grade = konversi_nilai(nilai)
        # tampilkan dengan 2 digit desimal biar rapi
        print(f"Nilai angka: {nilai:.2f} → Konversi huruf: {grade}")
    else:
        print("Nilai harus di antara 0 sampai 100!")
except ValueError:
    print("Input harus berupa angka (boleh desimal).")
```

# 6. While Loop

```python
import pandas as pd

# Inisialisasi nilai awal
nilai = 20
hasil = []

# While loop dengan increment 5
while nilai <= 100:
    hasil.append(nilai)
    nilai += 5

# Buat DataFrame untuk menampilkan hasil
df = pd.DataFrame({
    "Iterasi ke-": range(1, len(hasil)+1),
    "Nilai": hasil
})

print("Hasil looping dengan increment 5:")
display(df)  # kalau di Jupyter/Colab akan tampil tabel rapi
```
